Parameter What it Means

sentences The tokenized training corpus

vector_size=100 Each word will be represented by a 100-dimensional dense vector

window=2 Context window: how many words to the left/right of the target to consider

min_count=1 Ignore words with total frequency <1 (here, include all words)

sg=1 Use the Skip-Gram model (instead of CBOW)


```python
from gensim.models import Word2Vec

# Tokenized corpus
sentences = [
    ["We", "love", "natural", "language", "processing"],
    ["natural", "language", "processing", "is", "fun"],
    ["I", "enjoy", "machine", "learning"],
]

# Train Word2Vec model
model = Word2Vec(
    sentences, vector_size=100, window=4, min_count=1, sg=1
)  # sg=1 → skip-gram

# Get vector for a word
vector = model.wv["language"]

# Find similar words
model.wv.most_similar("language")
```
